```python
def simular_fifo(paginas, quantidade_frames):
    memoria = []
    faltas = 0
    hits = 0

    print("\fifo")
    for pagina in paginas:
        if pagina in memoria:
            hits += 1
            print(f"Página {pagina} -> {memoria} | HIT")
        else:
            faltas += 1
            if len(memoria) >= quantidade_frames:
                memoria.pop(0) # Remove o mais antigo
            memoria.append(pagina)
            print(f"Página {pagina} -> {memoria} | PAGE FAULT")

    return faltas, hits

def simular_lru(paginas, quantidade_frames):
    memoria = []
    faltas = 0
    hits = 0

    print("lru")
    for pagina in paginas:
        if pagina in memoria:
            hits += 1
            memoria.remove(pagina)
            memoria.append(pagina)
            print(f"Página {pagina} -> {memoria} | HIT")
        else:
            faltas += 1
            if len(memoria) >= quantidade_frames:
                memoria.pop(0)
            memoria.append(pagina)
            print(f"Página {pagina} -> {memoria} | PAGE FAULT")

    return faltas, hits

entrada = input("Páginas (separadas por espaço): ")
paginas_usuario = [int(valor) for valor in entrada.split()]
quantidade_frames = int(input("Quantidade de frames: "))

total_acessos = len(paginas_usuario)

# 1
faltas_fifo, hits_fifo = simular_fifo(paginas_usuario, quantidade_frames)

# 2
faltas_lru, hits_lru = simular_lru(paginas_usuario, quantidade_frames)

# 3,4,5
print("**********")
print("FIFO:")
print(f"Page Faults: {faltas_fifo} ({(faltas_fifo/total_acessos)*100:.1f}%)")
print(f"Hits: {hits_fifo} ({(hits_fifo/total_acessos)*100:.1f}%)")

print("\nLRU:")
print(f"Page Faults: {faltas_lru} ({(faltas_lru/total_acessos)*100:.1f}%)")
print(f"Hits: {hits_lru} ({(hits_lru/total_acessos)*100:.1f}%)")

if faltas_fifo < faltas_lru:
    print("\nMelhor desempenho: FIFO")
elif faltas_lru < faltas_fifo:
    print("\nMelhor desempenho: LRU")
else:
    print("\nEmpate: Deu zebra")
```

1- HIT é quando a página já está na RAM. PAGE FAULT é quando não está, aí o SO tem que ir buscar no SSD/HD, o que é mais lento.

2- O FIFO tira a página que entrou primeiro (tipo uma fila normal). O LRU tira a página que nós estamos há mais tempo sem acessar (tipo aba esquecida no navegador).

3- Não é erro de código. O nome "fault" é o processo normal do SO liberando espaço quando a RAM enche.

4- Não. por conta da Anomalia de Belady, o FIFO pode sofrer um aumento de page faults em certas sequências mesmo com mais frames.

5- Eles diferem quando páginas antigas são muito acessadas. O LRU mantém elas, mas o FIFO descarta.

6- Porque ele toma a decisão perfeita ao remover exatamente a página que vai demorar mais tempo para ser solicitada no futuro.

7- Porque o sistema operacional não tem a capacidade de prever exatamente quais páginas o programa vai pedir.

8- Executando os scripts, em sequências curtas os dois empatam, mas nos outros casos (anomalia de belady e sequências com muitas repetições) o LRU tem o melhor desempenho
